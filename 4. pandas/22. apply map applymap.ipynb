```python
# Series.map
import pandas as pd

# 使用函数
s = pd.Series([1, 2, 3, 4, 5])
print(s.map(lambda x:x*2))

# 使用字典映射
mapping = {1: "A", 2: "B", 3: "C", 4: "D", 5: "E"}
print(s.map(mapping))

# 使用 Series 映射
mapping_series = pd.Series(["A", "B", "C", "D", "E"], index=[1, 2, 3, 4, 5])
print(s.map(mapping_series))
```

```python
# 处理缺失值
import pandas as pd
import numpy as np

s = pd.Series([1, 2, np.nan, 4, 5])
print(s.map(lambda x:x*2 if pd.notna(x) else -1))
```

```python
# DataFrame.applymap
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [4, 5, 6],
    "C": [7, 8, 9]
})

# 对每个元素乘以2
print(df.map(lambda x:x*2))

# 保留2位小数
print(df.map(lambda x:round(x,2)))
```

```python
# DataFrame.apply
# apply 是最灵活的方法，可以沿着坐标轴应用函数。

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, 3, 4, 5],
    "B": [10, 20, 30, 40, 50],
    "C": [100, 200, 300, 400, 500]
})

# 默认 axis=0，按列应用
# 每列求和
print(df.apply(sum))

#每列最大值
print(df.apply(max))

# 按行应用
print(df.apply(sum,axis=1))
# 每行最大值-最小值
print(df.apply(lambda x:x.max()-x.min(),axis=1))
```

```python
# 使用 aggfunc 聚合
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [10, 20, 30]
})

# 同时应用多个函数
print(df.apply([sum,np.mean]))

# 返回多个值
r=df.apply(lambda x:pd.Series({
    "sum":x.sum(),
    "mean":x.mean(),
    "max":x.max()
},index=["sum","mean","max"]))
print(r)
```

```python
# Series.apply
import pandas as pd
import numpy as np

s = pd.Series([1, 4, 9, 16, 25])

# 开平方
print(s.apply(np.sqrt))

# 条件返回
print(s.apply(lambda x:">" if x>10 else "<"))
```
