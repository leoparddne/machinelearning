```python
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "张三", "李四", "赵六"],
    "年龄": [25, 30, 28, 25, 30, 35],
    "城市": ["北京", "上海", "广州", "北京", "深圳", "北京"]
})

# 检测重复行（默认从第一条开始标记，keep='first'）
print(df.duplicated())

# 统计重复行数量
print(df.duplicated().sum())

# 检测特定列的重复
print(df.duplicated(subset=["姓名"]))
```

```python
# keep 参数
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "张三", "王五", "李四"],
    "年龄": [25, 30, 25, 28, 30]
})

# keep='first'：保留第一次出现的（默认）
print(df.duplicated(keep="first"))

# keep='last'：保留最后一次出现的
print(df.duplicated(keep="last"))

# keep=False：标记所有重复项（不保留任何）
print(df.duplicated(keep=False))
```

```python
# 删除重复数据
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "张三", "王五", "李四"],
    "年龄": [25, 30, 25, 28, 30],
    "城市": ["北京", "上海", "北京", "广州", "深圳"]
})

print(df.drop_duplicates())
print(df.drop_duplicates(keep="last"))
print(df.drop_duplicates(keep=False))
```

```python
# 按列删除重复
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "张三", "王五"],
    "年龄": [25, 30, 35, 28],
    "城市": ["北京", "上海", "北京", "广州"]
})

# 按单列去重
print(df.drop_duplicates(subset=["姓名"]))

# 按多列组合去重
print(df.drop_duplicates(subset=["姓名","城市"]))

# 保留特定列值最大的行
print(df.sort_values("年龄",ascending=False).drop_duplicates(subset=["姓名"],keep="first"))
```
