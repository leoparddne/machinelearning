```python
# merge
import pandas as pd

# 示例数据
left = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
right = pd.DataFrame({'ID': [1, 2, 4], 'Age': [24, 27, 22]})

# 内连接
result=pd.merge(left,right,on="ID",how="inner")
print(result)
```

```python
# concat 沿轴连接
import pandas as pd

# 示例数据
df1 = pd.DataFrame({'A': [1, 2, 3]})
df2 = pd.DataFrame({'A': [4, 5, 6]})

result=pd.concat([df1,df2],axis=0,ignore_index=True)
print(result)
```

```python
# join 基于索引连接
import pandas as pd

# 示例数据
left = pd.DataFrame({'A': [1, 2, 3]}, index=[1, 2, 3])
right = pd.DataFrame({'B': [4, 5, 6]}, index=[1, 2, 4])

result=left.join(right,how="inner")
print(result)
```

```python
# 透视表
import pandas as pd

# 示例数据
data = {'Date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04'],
        'Category': ['A', 'B', 'A', 'B'],
        'Sales': [100, 150, 200, 250]}
df = pd.DataFrame(data)

pivot_table=pd.pivot_table(df,values="Sales",index="Date",columns="Category",aggfunc="sum",fill_value=0)
print(pivot_table)
```

```python
# 交叉表
import pandas as pd

# 示例数据
data = {'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
        'Region': ['North', 'South', 'North', 'South', 'West', 'East']}
df = pd.DataFrame(data)

cross_table=pd.crosstab(df["Category"],df["Region"])
print(cross_table)
```

```python
# apply 自定义函数
import pandas as pd

# 示例数据
df = pd.DataFrame({'A': [1, 2, 3, 4], 'B': [10, 20, 30, 40]})

def custom_func(x):
    return x*2

df['A']=df['A'].apply(custom_func)
print(df)

df = df.apply(lambda x: x ** 2)
print(df)
```

```python
# map 应用到series
df = pd.DataFrame({'A': ['cat', 'dog', 'rabbit']})
df['A']=df['A'].map({'cat':'kitten','dog':'puppy'})
print(df)
```

```python
# 分组
import pandas as pd

# 示例数据
df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value': [10, 20, 30, 40, 50, 60]
})

g=df.groupby('Category')['Value'].sum()
print(g)
```

```python
# agg 聚合
import pandas as pd

# 示例数据
df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value': [10, 20, 30, 40, 50, 60]
})

# 使用 agg() 来进行多个聚合操作
grouped = df.groupby('Category')['Value'].agg([sum,min,max])
print(grouped)
```

```python
# 时间的加减
import pandas as pd

# 日期偏移
date = pd.to_datetime('2024-01-01')
new_date = date + pd.Timedelta(days=10)
print(new_date)
```

```python
# 时间窗口
import pandas as pd

# 示例数据
df = pd.DataFrame({'Value': [10, 20, 30, 40, 50]})

# 计算 3 天滚动平均
df['Rolling_Mean'] = df['Value'].rolling(window=3).mean()
print(df)
```

```python
# 多重索引
import pandas as pd

# 创建元组
index_tuples = [('A', 1), ('A', 2), ('B', 1), ('B', 2)]

multi_index=pd.MultiIndex.from_tuples(index_tuples,names=['Letter','Number'])

df=pd.DataFrame({'Value':[10,20,30,40]},index=multi_index)
print(df)
```

```python
# from_product 使用多个列表的笛卡尔积来创建多重索引
import pandas as pd

# 创建多个列表
index_values = [['A', 'B'], [1, 2]]
multi_index=pd.MultiIndex.from_product(index_values,names=['Letter','Number'])

df=pd.DataFrame({'Value':[10,20,30,40]},index=multi_index)
print(df)
```

```python
# set_index 创建多重索引
import pandas as pd

# 示例数据
data = {
    'Letter': ['A', 'A', 'B', 'B'],
    'Number': [1, 2, 1, 2],
    'Value': [10, 20, 30, 40]
}

df = pd.DataFrame(data)

# 设置多重索引
df.set_index(['Letter','Number'],inplace=True)
print(df)
```

```python
# 访问多重索引数据
import pandas as pd

# 示例数据
data = {
    'Letter': ['A', 'A', 'B', 'B'],
    'Number': [1, 2, 1, 2],
    'Value': [10, 20, 30, 40]
}

df = pd.DataFrame(data)

# 设置多重索引
df.set_index(['Letter', 'Number'], inplace=True)
print(df.loc['A',1])

# 使用 xs() 获取交叉数据
print(df.xs(1,level='Number'))
```

```python
# 多重索引的切片
# 选择 Letter 为 'A' 的所有数据
print(df.loc['A'])
```

```python
# 排序多重索引
df_sorted=df.sort_index(level=['Letter','Number'],ascending=[True,False])
print(df_sorted)
```

```python
# 多重索引聚合操作
df_group=df.groupby(['Letter','Number']).sum()
print(df_group)
```

```python
# 重设索引
df_reset=df.reset_index()
print(df_reset)
```

```python
# 多重索引的缺失值
data = {
    'Letter': ['A', 'A', 'B', 'B'],
    'Number': [1, 2, 1, 2],
    'Value': [10, None, 30, 40]
}

df = pd.DataFrame(data)
df.set_index(['Letter', 'Number'], inplace=True)

df_fill=df.fillna(0)
print(df_fill)
```
