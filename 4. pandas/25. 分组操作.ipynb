```python
import pandas as pd

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营", "销售", "技术"],
    "姓名": ["张三", "李四", "王五", "赵六", "钱七", "孙八"],
    "薪资": [12000, 15000, 11000, 18000, 14000, 13000]
})

g=df.groupby("部门")

for name,group in g:
    print(f"{name},{group}")
```

```python
# 分组并聚合
import pandas as pd

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营", "销售"],
    "薪资": [12000, 15000, 11000, 18000, 14000]
})

print(df.groupby("部门")["薪资"].sum())
```

```python
# 多列分组
import pandas as pd

df = pd.DataFrame({
    "年份": ["2023", "2023", "2024", "2024"],
    "部门": ["技术", "销售", "技术", "销售"],
    "薪资": [12000, 15000, 13000, 16000]
})

# 按年份和部门分组
print(df.groupby(["年份","部门"])["薪资"].sum())

# 作为 DataFrame 输出
result=df.groupby(["年份","部门"]).agg({
    "薪资":"sum"
}).reset_index()
print(result)
```

```python
# 单一聚合
import pandas as pd

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营"],
    "薪资": [12000, 15000, 11000, 18000]
})

# 链式调用多个聚合
result=df.groupby("部门")["薪资"].agg(["sum","mean","max","min"])
print(result)
```

```python
# 自定义聚合
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营"],
    "薪资": [12000, 15000, 11000, 18000]
})

# 自定义聚合函数
def range_func(x):
    return x.max()-x.min()

print(df.groupby("部门")["薪资"].agg(range_func))

# 命名聚合（推荐）
print(df.groupby("部门").agg(
    最低薪资=("薪资","min"),
    最高薪资=("薪资","max"),
    平均薪资=("薪资","mean"),
))
```

```python
# transform
# 在保持原数据形状的同时，对每个分组进行计算并返回与原数据相同形状的结果
import pandas as pd

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营", "销售"],
    "薪资": [12000, 15000, 11000, 18000, 14000]
})

df["mean"]=df.groupby("部门")["薪资"].transform("mean")
print(df)
```

```python
# filter 过滤
import pandas as pd

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营", "销售", "技术"],
    "薪资": [12000, 15000, 11000, 18000, 14000, 13000]
})

filtered=df.groupby("部门").filter(lambda x:x["薪资"].mean()>13000)
print(filtered)

filtered=df.groupby("部门").filter(lambda x:len(x)>2)
print(filtered)
```

```python
# apply 自定义函数
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "部门": ["技术", "销售", "技术", "运营"],
    "薪资": [12000, 15000, 11000, 18000]
})

result=df.groupby("部门").apply(
    lambda x:pd.Series({
        "总和":x["薪资"].sum(),
        "均值":x["薪资"].mean(),
    })
)
print(result)
```
