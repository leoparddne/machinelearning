```python
# 基础过滤
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六", "钱七"],
    "年龄": [25, 30, 28, 35, 22],
    "城市": ["北京", "上海", "广州", "北京", "深圳"],
    "薪资": [12000, 15000, 11000, 18000, 9000],
    "部门": ["技术", "销售", "技术", "运营", "技术"]
})
print(df)

print(df[df["城市"]=="北京"])
print(df[df["城市"]!="北京"])
print(df[df["薪资"]>10000])

# 字符串条件
print(df[df["姓名"].str.contains("张")])
```

```python
# 多条件组合
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六", "钱七"],
    "年龄": [25, 30, 28, 35, 22],
    "城市": ["北京", "上海", "广州", "北京", "深圳"],
    "薪资": [12000, 15000, 11000, 18000, 9000],
    "部门": ["技术", "销售", "技术", "运营", "技术"]
})

print(df[(df["城市"]=="北京")&(df["部门"]=="技术")])
print(df[(df["城市"]=="北京")|(df["部门"]=="技术")])
print(df[~(df["城市"]=="北京")])
```

```python
# isin 与 where
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六", "钱七"],
    "城市": ["北京", "上海", "广州", "北京", "深圳"],
    "部门": ["技术", "销售", "技术", "运营", "技术"]
})

print(df[df["城市"].isin(["北京","上海"])])
print(df[~df["城市"].isin(["北京","上海"])])

# 多列 isin
df2 = pd.DataFrame({
    "城市": ["北京", "上海"],
    "部门": ["技术", "销售"]
})
print(df[df[["城市","部门"]].isin(df2).all(axis=1)])
```

```python
# where
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 5, 3, 7],
    "B": [4, 2, 8, 3]
})

# 值大于3的保留，其他设为0
result=df.where(df>3,0)
print(result)

# 使用其他 DataFrame 作为条件
other = pd.DataFrame({
    "A": [True, False, True, False],
    "B": [False, True, False, True]
})
print(df.where(other,-1))
```

```python
# 字符串过滤
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六", "孙七"],
    "城市": ["北京市", "上海市", "广州市", "深圳市", "杭州市"],
    "邮箱": ["zhangsan@email.com", "lisi@email.com", "wangwu@email.com",
             "zhaoliu@email.com", "sunqi@email.com"]
})

# 包含
print(df[df["城市"].str.contains("北京")])

# 开头/结尾
print(df[df["城市"].str.startswith("北京")])

# 正则
print(df[df["邮箱"].str.extract(r"(\w+)@")[0].str.len()>4])

# 多个模式匹配
print(df[df["城市"].str.contains("^北|^上|^广|^深")])
```

```python
# 数值范围过滤
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六", "钱七"],
    "年龄": [25, 30, 28, 35, 22],
    "薪资": [12000, 15000, 11000, 18000, 9000]
})

# 范围过滤
print(df[df["年龄"].between(25,30)])

# 分位数筛选
q1=df["薪资"].quantile(0.25)
q3=df["薪资"].quantile(0.75)
print(df[df["薪资"].between(q1,q3)])

# 最小/最大N条
print(df.nlargest(3,"薪资"))
print(df.nsmallest(3,"薪资"))
```

```python
# 时间数据过滤
import pandas as pd

# 创建时间序列数据
df = pd.DataFrame({
    "日期": pd.date_range("2024-01-01", periods=10, freq="D"),
    "销量": [100, 120, 90, 150, 200, 180, 110, 130, 170, 190]
})
df = df.set_index("日期")
# print(df)

# 按日期筛选
print(df.loc["2024-01-01"])

# 日期范围筛选
print(df.loc["2024-01-01":"2025-01-01"])

# 使用 truncate（更高效的切片）
print(df.truncate(before="2024-01-03",after="2024-01-07"))
```

```python
# query 方法
# 更简洁的 SQL 风格的过滤方式
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六"],
    "年龄": [25, 30, 28, 35],
    "城市": ["北京", "上海", "广州", "北京"],
    "薪资": [12000, 15000, 11000, 18000]
})
print(df.query("年龄>26 and 城市=='北京'"))

# 使用变量
min_age = 26
city = "北京"
print(df.query("年龄>@min_age and 城市==@city"))

# 列名有空格时需要用反引号
df2=df.rename(columns={"姓名":"姓 名"})
print(df2.query("`姓 名`=='张三'"))
```
