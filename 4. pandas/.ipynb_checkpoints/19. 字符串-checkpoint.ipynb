```python
import pandas as pd

s = pd.Series(["hello", "world", "pandas", "Python"])

print(s.str.upper())
print(s.str.lower())
print(s.str.capitalize())

```

```python
import pandas as pd

s = pd.Series(["hello world", "python pandas", "data science", "machine learning"])

print(s.str.contains("Python",case=False))
print(s.str.startswith("Python"))
print(s.str.replace("Python","xx"))
print(s.str.split(" "))
```

```python
import pandas as pd

s = pd.Series(["  hello  ", "world  ", "  pandas", " python "])

# 去除首尾空格
print(s.str.strip())
print(s.str.lstrip())
print(s.str.rstrip())

# 填充
print(s.str.pad(10,side="left",fillchar="0"))
```

```python
import pandas as pd

s = pd.Series(["hello", "world", "python", "pandas"])

# 切片
print(s.str[:3])
print(s.str[1:4])

# 字符串连接
s1 = pd.Series(["Hello", "World"])
s2 = pd.Series(["Python", "Pandas"])
print(s1+s2)

# join 连接
print(s1.str.cat(s2,sep="-"))
```

```python
# 正则表达式
import pandas as pd

s = pd.Series([
    "user@example.com",
    "test@domain.org",
    "invalid-email",
    "admin@site.net"
])

# 提取邮箱
print(s.str.extract(r"(\w+)@"))

# 提取域名
print(s.str.extract(r"@(\w+\.\w+)"))

# 替换
print(s.str.replace(r"\w+@\w+\.\w+","[email]",regex=True))

# 包含匹配
print(s.str.contains(r"\.com|\.org"))
```

```python
# 提取和分割
import pandas as pd

s = pd.Series(["2024-01-01", "2024-02-15", "2024-03-20"])

# 提取年月日
df=s.str.extract(r"(\d{4})-(\d{2})-(\d{2})")
df.columns=["年","月","日"]
print(df)

# 按分隔符拆分
s2 = pd.Series(["a,b,c", "d,e,f", "g,h,i"])
print(s2.str.split(","))

# 拆分并展开为 DataFrame
print(s2.str.split(",",expand=True))
```

```python
# 长度与计数
import pandas as pd

s = pd.Series(["hello", "world", "python", "pandas"])

# 字符串长度
print(s.str.len())

# 字符计数
s2 = pd.Series(["hello", "world", "aaa", "bbb"])
# 字符'l'出现次数
print(s2.str.count("l"))
```

```python
# 类型转换
import pandas as pd

s = pd.Series(["1", "2", "3", "abc"])

# 字符串转数值
print(pd.to_numeric(s, errors="coerce"))

# 数值转字符串
s2 = pd.Series([1, 2, 3, 4])
# 补零到4位
print(s2.astype(str).str.zfill(4))
```
