```python
import pandas as pd
import numpy as np

# 创建各种数据类型的 DataFrame
df = pd.DataFrame({
    "整数": [1, 2, 3],
    "浮点数": [1.5, 2.5, 3.5],
    "字符串": ["a", "b", "c"],
    "布尔值": [True, False, True],
    "日期": pd.date_range("2024-01-01", periods=3)
})

print("各列数据类型：")
print(df.dtypes)
```

```python
# 自动类型推断
import pandas as pd

# CSV 读取时的类型推断
# pandas 会尝试推断每列的最合适类型
df = pd.read_csv("data.csv")
print(df.dtypes)

# 或者使用 dtype 参数显式指定类型
df = pd.read_csv("data.csv", dtype={
    "年龄": "int32",     # 指定为32位整数，节省内存
    "薪资": "float32",   # 指定为32位浮点数
    "姓名": "string"     # 使用 PyArrow 字符串
})
```

```python
# 创建时指定类型
import pandas as pd
import numpy as np

# 指定数据类型创建 Series
s = pd.Series([1, 2, 3], dtype="int8")  # 使用更小的整数类型
print(s.dtype)

s = pd.Series([1.5, 2.5, 3.5], dtype="float32")  # 使用 float32
print(s.dtype)

# 使用 numpy 类型
s = pd.Series([1, 2, 3], dtype=np.int8)
print(s.dtype)
```

```python
# 类型转换
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "整数": [1, 2, 3],
    "浮点数": [1.5, 2.5, 3.5],
    "字符串": ["1", "2", "3"],
    "布尔值": [1, 0, 1]
})

# 字符串转数值
df["整数_str"] = df["整数"].astype(str)
print(df.dtypes)

# 数值转布尔值（非0为True）
df["布尔_int"] = df["布尔值"].astype(bool)
print(df.dtypes)

df["浮点_int"] = df["浮点数"].astype(int)
```

```python
# 使用 pd.to_numeric 转换

import pandas as pd
import numpy as np

# 处理带特殊字符的数值字符串
s = pd.Series(["$1,000", "$2,500", "$3,200"])

# 清理并转换
clean=s.str.replace("$","",regex=False).str.replace(",","",regex=False)
num=pd.to_numeric(clean)
print(num)

# 处理缺失值
s_with_na = pd.Series(["1", "2", "NA", "4"])
# 无效值转为 NaN
s_num=pd.to_numeric(s_with_na,errors="coerce")
print(s_num)
```

```python
# pd.to_datetime 转换日期
import pandas as pd

# 各种日期格式转换
dates = ["2024-01-01", "2024/01/02", "01/03/2024", "20240104"]

dt=pd.to_datetime(dates,errors="coerce")
print(dt)

# 指定日期格式
dates2 = ["20240101", "20240102", "20240103"]
dt2 = pd.to_datetime(dates2,format="%Y%m%d")
print(dt2)
```

```python
# 类型检查与判断
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "整数": [1, 2, 3],
    "浮点": [1.5, 2.5, 3.5],
    "字符串": ["a", "b", "c"]
})

# 使用 is_integer_dtype
print(pd.api.types.is_integer_dtype(df['整数']))

# 检查是否可以转换为数值
print(pd.api.types.is_numeric_dtype(df['整数']))
```
