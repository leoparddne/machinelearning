```python
# 处理有限类别值的数据类型，特别适合处理枚举类型的数据
import pandas as pd

# 创建类别数据
s = pd.Series(["男", "女", "男", "女", "男"],dtype="category")
print(s)

# 指定类别顺序
s2 = pd.Series(
    ["低", "中", "高", "中", "低"],
    dtype=pd.CategoricalDtype(categories=["低", "中", "高"],ordered=True))

print(s2)
```

```python
# 从列表创建
import pandas as pd

categories = pd.Categorical(
    ["甲", "乙", "甲", "丙", "乙"],
    categories=["甲", "乙", "丙"]
)

s=pd.Series(categories)
print(s)
```

```python
# 访问类别信息
import pandas as pd

s = pd.Series(
    ["低", "中", "高", "中", "低"],
    dtype=pd.CategoricalDtype(categories=["低", "中", "高"], ordered=True)
)

print(s.value_counts())
```

```python
# 修改类别
import pandas as pd

s = pd.Series(["A", "B", "C", "A", "B"], dtype="category")
print(s)

s2=s.cat.rename_categories({"A": "优秀", "B": "良好", "C": "及格"})
print(s2)

# 添加新类别
s3=s.cat.add_categories(["D"])
print(s3)
print(s3.dtype.categories.tolist())

# 删除类别
s4=s.cat.remove_categories(["C"])
print(s4.dtype.categories.tolist())
```

```python
# 有序类别
# 有序类别支持比较操作，适用于有大小顺序的分类数据。

import pandas as pd

# 创建有序类别
s = pd.Series(
    ["低", "中", "高", "高", "低", "中"],
    dtype=pd.CategoricalDtype(categories=["低", "中", "高"], ordered=True)
)

# 比较操作
print(s>"中")

# 排序
print(s.sort_values())
```
