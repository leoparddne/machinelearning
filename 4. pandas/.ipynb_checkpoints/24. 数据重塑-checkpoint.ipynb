```python
# pivot 透视表
import pandas as pd

df = pd.DataFrame({
    "日期": ["2024-01-01", "2024-01-01", "2024-01-02", "2024-01-02"],
    "产品": ["A", "B", "A", "B"],
    "销量": [100, 150, 120, 90]
})

pivot_df=df.pivot(index="日期",columns="产品",values="销量")
print(pivot_df)
```

```python
# 多重索引
import pandas as pd

df = pd.DataFrame({
    "年份": ["2024", "2024", "2024", "2024"],
    "季度": ["Q1", "Q1", "Q2", "Q2"],
    "产品": ["A", "B", "A", "B"],
    "销量": [100, 150, 120, 90]
})

pivot_df=df.pivot(index=["年份","季度"],columns="产品",values="销量")
print(pivot_df)
```

```python
# 聚合函数
# pivot 不允许有重复值，有重复值时使用 pivot_table 并指定聚合函数。

import pandas as pd

df = pd.DataFrame({
    "日期": ["2024-01-01", "2024-01-01", "2024-01-01", "2024-01-02"],
    "产品": ["A", "A", "B", "B"],
    "销量": [100, 110, 80, 90]
})

# pivot 默认不支持重复值，需要用 pivot_table 并指定聚合函数
pivot_df=df.pivot_table(index="日期",columns="产品",values="销量",aggfunc="mean")
print(pivot_df)
```

```python
# melt 逆透视
# melt 将宽格式数据转换为长格式，是 pivot 的逆操作。

import pandas as pd

# 创建宽格式数据
df = pd.DataFrame({
    "日期": ["2024-01-01", "2024-01-02"],
    "A": [100, 120],
    "B": [150, 90]
})

melted=df.melt(id_vars="日期",var_name="产品",value_name="销量")

print(melted)
```

```python
# 保持多列不变
import pandas as pd

df = pd.DataFrame({
    "城市": ["北京", "上海"],
    "2023营收": [1000, 800],
    "2023利润": [200, 150],
    "2024营收": [1200, 950],
    "2024利润": [250, 180]
})

# 保持"城市"不变，其他列转换为长格式
melted=df.melt(id_vars="城市",var_name="指标",value_name="数值")

print(melted)
```

```python
# stack 与 unstack
# stack 和 unstack 是 MultiIndex 专用的重塑函数。

# 创建带多层索引的数据
df = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": [5, 6, 7, 8]
}, index=pd.MultiIndex.from_tuples(
    [("X", 1), ("X", 2), ("Y", 1), ("Y", 2)],
    names=["类别", "编号"]
))

print(df.unstack())

df = pd.DataFrame({
    ("A", "X"): [1, 2],
    ("A", "Y"): [3, 4],
    ("B", "X"): [5, 6],
    ("B", "Y"): [7, 8]
})
print(df.stack())
```
