```python
# 自定义列名并跳过前两行

df=pd.read_excel("data.xlsx",header=None,names=['a','b','c','d','e'],skiprows=2)
print(df)
```

```python
# 写入excel
df = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie'],
'Age': [25, 30, 35],
'City': ['New York', 'Los Angeles', 'Chicago']
})

df.to_excel("out.xlsx",sheet_name="sheet1",index=False)

df2 = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie'],
'Age': [25, 30, 35],
'City': ['New York', 'Los Angeles', 'Chicago']
})

# 写入多个表单
with pd.ExcelWriter("out2.xlsx") as writer:
    df.to_excel(writer,sheet_name="sheet1",index=False)
    df2.to_excel(writer,sheet_name="sheet2",index=False)
```

```python
# 查看所有表单名称
excelFile=pd.ExcelFile("data.xlsx")
print(excelFile.sheet_names)

# 读取表单
df=excelFile.parse("data")
print(df)

# 关闭文件
excelFile.close()
```

```python
# 设置时间序列化格式
from datetime import date,datetime
df = pd.DataFrame(
    [
        [date(2014, 1, 31), date(1999, 9, 24)],
        [datetime(1998, 5, 26, 23, 33, 4), datetime(2014, 2, 28, 13, 5, 13)],
    ],
    index=["Date", "Datetime"],
    columns=["X", "Y"],
) 

with pd.ExcelWriter("date.xlsx",date_format="YYYY-MM-DD",datetime_format="YYYY-MM-DD HH:MM:SS") as writer:
    df.to_excel(writer)
```

```python
# 追加到excel
import pandas as pd

df = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie'],
'Age': [25, 30, 35],
'City': ['New York', 'Los Angeles', 'Chicago']
})

# 多次写入同一个sheet需要曾敏if_sheet_exists参数
with pd.ExcelWriter("data.xlsx",mode="a",engine="openpyxl",if_sheet_exists="new") as writer:
    df.to_excel(writer,sheet_name="sheet1")
```

```python
# 保存在内存中便于直接传输,不产生中间文件
import io
df = pd.DataFrame([["ABC", "XYZ"]], columns=["Foo", "Bar"])
buffer=io.BytesIO()
with pd.ExcelWriter(buffer) as writer:
    df.to_excel(writer)


# 读取
# 获取二进制字节
buffer.seek(0)
d=buffer.getvalue()
print(len(d))


# 保存到硬盘（如果需要的话）
with open("output.xlsx", "wb") as f:
    f.write(d)
```

```python
# 将 Excel 文件打包到 zip 压缩文件中
import zipfile
df = pd.DataFrame([["ABC", "XYZ"]], columns=["Foo", "Bar"])  

with zipfile.ZipFile("a.zip","w") as zf:
    with zf.open("file.xlsx","w")as buffer:
        with pd.ExcelWriter(buffer) as writer:
            df.to_excel(writer)
```
