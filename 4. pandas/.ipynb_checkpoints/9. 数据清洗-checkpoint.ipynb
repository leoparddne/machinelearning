```python
import pandas as pd

df=pd.read_csv("property-data.csv")
print(df)

print(df['NUM_BEDROOMS'])
print(df['NUM_BEDROOMS'].isnull())
```

```python
# 指定空数据类型
import pandas as pd
missing_values=["n/a","na","--"]

df=pd.read_csv("property-data.csv",na_values=missing_values)

print(df['NUM_BEDROOMS'])
print(df['NUM_BEDROOMS'].isnull())
```

```python
# 删除包含空数据的行
import pandas as pd

df=pd.read_csv("property-data.csv")
data=df.dropna()
print(data)
```

```python
# dropna默认情况下返回修改后的数据而不是原地删除
# 可以通过inplace调整行为
df=pd.read_csv("property-data.csv")
df.dropna(inplace=True)
print(df)
```

```python
# 移除指定列有空行的数据
import pandas as pd

df=pd.read_csv("property-data.csv")
df.dropna(subset=['ST_NUM'],inplace=True)
print(df)
```

```python
# 替换空字符
import pandas as pd

df = pd.read_csv('property-data.csv')

# 对不同类型列分别填充
df['ST_NAME'] = df['OWN_OCCUPIED'].fillna('未知')      # 字符串列填文本
df['PID'].fillna(12345)

print(df)
```

```python
# 使用均值替换空单元格
import pandas as pd

df = pd.read_csv('property-data.csv')

# 均值
x = df["ST_NUM"].mean()
# 中位数
x = df["ST_NUM"].median()
# 众数
x = df["ST_NUM"].mode()

df['ST_NUM']=df['ST_NUM'].fillna(x)

print(df)
```

```python
# 清洗错误格式数据
import pandas as pd

# 第三个日期格式错误
data = {
  "Date": ['2020/12/01', '2020/12/02' , '20201226'],
  "duration": [50, 40, 45]
}

df=pd.DataFrame(data,index=["d1","d2","d3"])

df['Data']=pd.to_datetime(df['Date'],format='mixed')
print(df)
```

```python
# 清洗错误数据
import pandas as pd

person = {
  "name": ['Google', 'Runoob' , 'Taobao'],
  "age": [50, 40, 12345]    # 12345 年龄数据是错误的
}

df=pd.DataFrame(person)

df.loc[2,'age']=30
print(df)
```

```python
# 条件语句修改数据
import pandas as pd

person = {
  "name": ['Google', 'Runoob' , 'Taobao'],
  "age": [50, 200, 12345]    
}

df=pd.DataFrame(person)

for x in df.index:
    if df.loc[x,"age"]>120:
        df.loc[x,"age"]=120

print(df)
```

```python
# 条件语句删除指定行
import pandas as pd

person = {
  "name": ['Google', 'Runoob' , 'Taobao'],
  "age": [50, 200, 12345]    
}

df=pd.DataFrame(person)

for x in df.index:
    if df.loc[x,"age"]>120:
        df.drop(x,inplace=True)

print(df)
```

```python
# 清洗重复数据
import pandas as pd

person = {
  "name": ['Google', 'Runoob', 'Runoob', 'Taobao'],
  "age": [50, 40, 40, 23]  
}
df=pd.DataFrame(person)

print(df.duplicated())

# 删除重复数据
df.drop_duplicates(inplace=True)
print(df)
```

```python
# 独热编码
import pandas as pd

# 示例数据
data = {'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']}

df = pd.DataFrame(data)

df_encoded=pd.get_dummies(df,columns=['City'])
print(df_encoded)
```

```python
# 标准化
from sklearn.preprocessing import StandardScaler
import pandas as pd
data = {'Age': [25, 30, 35, 40, 45],
        'Salary': [50000, 60000, 70000, 80000, 90000]}

df=pd.DataFrame(data)

scaler=StandardScaler()

df_scaled=scaler.fit_transform(df)
print(df_scaled)
```
