```python
# 读取复杂结构
# 展平指定的json字段
with open("nested_mix.json",'r') as f:
    data=json.loads(f.read())
    
# meta展示其他字段
df=pd.json_normalize(data,record_path=['students'],meta=['class',['info','president'],['info','contacts','tel']])
print(df)
```

```python
# 访问内嵌的属性
import pandas as pd
from glom import glom

df=pd.read_json('nested_deep.json')
d=df['students'].apply(lambda row:glom(row,'grade.math'))
print(d)
```

```python
# dataframe转json
import pandas as pd

# 创建 DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
})

json=df.to_json()
print(json)
```

```python
# dataframe保存到json文件
import pandas as pd

# 创建 DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
})

df.to_json('save.json',orient='records',lines=True)
```

```python
# 保存json文件并指定日期格式
jsonstr=df.to_json(date_format='iso')
print(jsonstr)
```
