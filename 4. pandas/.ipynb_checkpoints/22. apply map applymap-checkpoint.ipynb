```python
# Series.map
import pandas as pd

# 使用函数
s = pd.Series([1, 2, 3, 4, 5])
print(s.map(lambda x:x*2))

# 使用字典映射
mapping = {1: "A", 2: "B", 3: "C", 4: "D", 5: "E"}
print(s.map(mapping))

# 使用 Series 映射
mapping_series = pd.Series(["A", "B", "C", "D", "E"], index=[1, 2, 3, 4, 5])
print(s.map(mapping_series))
```
