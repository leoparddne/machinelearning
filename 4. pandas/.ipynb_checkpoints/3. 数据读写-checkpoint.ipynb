```python
# csv
import pandas as pd

df =pd.read_csv("data.csv")
print(df)

# 指定分隔符
df_tsv=pd.read_csv("data.csv",sep="\t")

# 编码
df_utf8=pd.read_csv("data.csv",encoding="utf-8")
# df_gbk=pd.read_csv("data.csv",encoding="gbk")
```

```python
# 跳过前三行
df=pd.read_csv("data.csv",skiprows=3)
print(df)

# 跳过第2、4行
df=pd.read_csv("data.csv",skiprows=[2,4])
print(df)
```

```python
# 设置表头行
df=pd.read_csv("data.csv",header=0)
print(df)

# 不使用表头
df=pd.read_csv("data.csv",header=None)
print(df)
```

```python
# 指定列名
df=pd.read_csv("data.csv",header=0,names=["name","age","city","job","salary"])
print(df)
```

```python
# 缺失值处理
# 将特定值识别为Nanabs
df=pd.read_csv("data.csv",na_values=["NA","null","NULL"])
print(df)

# 不同列用不同的值
df=pd.read_csv("data.csv",na_values={"姓名":["未知","null","NULL"],"Age":["NA"]})
print(df)
```

```python
# 写入csv
df=pd.DataFrame({"Name":["a","b"],"Age":[22,23]})

# 写入 CSV（默认有索引）
df.to_csv("out.csv")

# 不写入索引
df.to_csv("out.csv",index=False)

# 指定分隔符
df.to_csv("out.csv",sep="\t")

# 不写入表头
df.to_csv("out.csv",header=False)

# 指定编码
df.to_csv("out.csv",encoding="utf-8")
```

```python
# json
import pandas as pd
import json
from io import StringIO

data_records = '''
[
    {"name": "张三", "age": 25, "city": "北京"},
    {"name": "李四", "age": 30, "city": "上海"},
    {"name": "王五", "age": 28, "city": "广州"}
]
'''

# json数组转DataFrame
df = pd.read_json(StringIO(data_records), orient="records")
print(df)

# json对象装series
data ='{"name": "张三", "age": 25, "city": "北京"}'
s=pd.read_json(StringIO(data),typ="series")
print(s)
```

```python
import pandas as pd

# JSON Lines 格式数据
jsonl_data = '''{"name": "张三", "age": 25}
{"name": "李四", "age": 30}
{"name": "王五", "age": 28}
{"name": "赵六", "age": 35}
'''

# 逐行写入json数据
with open("data.json","w",encoding="utf-8") as f:
    f.write(jsonl_data)

# 读取逐行的json
df=pd.read_json("data.json",lines=True)
print(df)
```

```python
# 写入json
df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"],
    "年龄": [25, 30, 28],
    "城市": ["北京", "上海", "广州"]
})
df.to_json("data1.json",orient="records",force_ascii=False,indent=2)

# 写入不同的orient
df.to_json("data1_index.json",orient="index",force_ascii=False,indent=2)
df.to_json("data1_columns.json",orient="columns",force_ascii=False,indent=2)

df.to_json("data1_lines.json",orient="records",lines=True,force_ascii=False,indent=2)
```

```python
# pickle序列化
# Python 原生的对象序列化格式，可以保存任何 Python 对象
import pandas as pd
import pickle

df = pd.DataFrame({
    "A": range(10),
    "B": range(10, 20),
    "C": ["foo", "bar"] * 5
})

df.to_pickle("data.pk1")

df_load=pd.read_pickle("data.pk1")
print(df_load)

# 压缩写入
df.to_pickle("data.pk1.gz",compression="gzip")
df_load=pd.read_pickle("data.pk1.gz",compression="gzip")
print(df_load)
```
