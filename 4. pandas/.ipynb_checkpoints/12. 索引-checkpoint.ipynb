```python
import pandas as pd

# 创建简单的 DataFrame（默认使用 RangeIndex）
df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"],
    "年龄": [25, 30, 28],
    "城市": ["北京", "上海", "广州"]
})

# 行索引
print(df.index.tolist())

# 列索引
print(df.columns.tolist())
```

```python
# rangeindex 
# RangeIndex 是默认的整数索引，类似 Python 的 range(n)，从 0 开始递增
import pandas as pd
idx=pd.RangeIndex(start=0,stop=10,step=1)
print(idx)
print(type(idx))

df = pd.DataFrame({"A": [1, 2, 3]}, index=range(3))
print(type(df.index))

df_reset=df.reset_index()
print(type(df_reset.index))
```

```python
# index 类型转换
import pandas as pd
df = pd.DataFrame({"值": [1, 2, 3, 4]}, index=[10, 20, 30, 40])

# tolist
idx_list=df.index.tolist()
print(type(idx_list))

# to array
idx_array=df.index.values
print(idx_array)
```

```python
# 自定义索引
import pandas as pd

df = pd.DataFrame({
    "学号": ["S001", "S002", "S003", "S004"],
    "姓名": ["张三", "李四", "王五", "赵六"],
    "分数": [85, 92, 78, 90]
})

# 将学号设置为索引
df1=df.set_index("学号")
print(df1)

# 设置多个索引
df2=df.set_index(["学号","姓名"])
print(df2)
```

```python
# 使用 index 参数指定索引
import pandas as pd
d=pd.date_range("2024-01-01",periods=3,freq="D")

df_data=pd.DataFrame({"v":[100,200,300]},index=d)

print(df_data)
```

```python
# 重置索引
import pandas as pd

df = pd.DataFrame(
    {"姓名": ["张三", "李四"], "分数": [85, 92]},
    index=["A001", "A002"]
)

df_reset =df.reset_index()
print(df)
print(df_reset)

# drop 参数：是否丢弃原索引列
df_reset2=df.reset_index(drop=True)
print(df_reset2)
```

```python
# 重新索引
import pandas as pd

df = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]}, index=[1, 2, 3])
reindex=df.reindex([1,2,3,4,5])
print(reindex)

# 填充缺失值
fe_reindex=df.reindex([1,2,3,4,5],fill_value=0)
print(fe_reindex)
```

```python
# 索引层级操作
import pandas as pd
df = pd.DataFrame({
    "语文": [85, 92, 78],
    "数学": [90, 88, 95]
},index=pd.MultiIndex.from_tuples([("高一", "A班"), ("高一", "B班"), ("高二", "A班")],names=["年级","班级"]))
print(df)

# 获取外层索引
print(df.index.get_level_values(0).tolist())
# 获取内层索引
print(df.index.get_level_values(1).tolist())
```

```python
# 索引属性
import pandas as pd

df = pd.DataFrame(
    {"值": [1, 2, 3, 4]},
    index=["a", "b", "c", "d"]
)

# 索引是否唯一
print(df.index.is_unique)

# 索引长度
print(len(df.index))
print(df.index.dtype)

# 转换为字符串类型
str_index = df.index.astype(str)
print(str_index.dtype)
```

```python
# 多层索引
import pandas as pd
arrays = [
    ["A", "A", "B", "B", "C", "C"],
    [1, 2, 1, 2, 1, 2]
]

index=pd.MultiIndex.from_arrays(arrays,names=["类别","编号"])
print(index)


# 使用元组列表
tuples = [
    ("A", 1), ("A", 2), ("B", 1), ("B", 2), ("C", 1), ("C", 2)
]
index=pd.MultiIndex.from_tuples(tuples,names=["类别","编号"])
print(index)


# 使用product
index=pd.MultiIndex.from_product([["A", "B", "C"], [1, 2, 3]],names=["类别","编号"])
print(index)
```

```python
# 创建带 MultiIndex 的 DataFrame

import pandas as pd
import numpy as np
df=pd.DataFrame(
    np.random.randn(6, 4),
    index=pd.MultiIndex.from_tuples([("2024", "Q1"), ("2024", "Q2"), ("2024", "Q3"),
         ("2025", "Q1"), ("2025", "Q2"), ("2025", "Q3")]
    ),
    columns=["北京", "上海", "广州", "深圳"]
)

print(df)
print(df.index)
df.index.names=["年份", "季度"]
print(df)
```

```python
# 多层索引的访问
# 使用 loc/iloc 访问
df = pd.DataFrame({
    "语文": [85, 92, 78, 88],
    "数学": [90, 88, 95, 82]
}, index=pd.MultiIndex.from_tuples(
    [("高一", "A班"), ("高一", "B班"), ("高二", "A班"), ("高二", "B班")],
    names=["年级", "班级"]
))

print(df)

# 外层索引访问
print(df.loc["高一"])

# 内层索引访问
# print(df.loc[:, "B班"])
print(df.xs("A班", level="班级"))


# 多层索引访问
print(df.loc[("高一", "A班")])
```

```python
# 使用 xs 访问
import pandas as pd

# 创建示例数据
df = pd.DataFrame({
    "语文": [85, 92, 78, 88],
    "数学": [90, 88, 95, 82]
}, index=pd.MultiIndex.from_tuples(
    [("高一", "A班"), ("高一", "B班"), ("高二", "A班"), ("高二", "B班")],
    names=["年级", "班级"]
))

print(df.xs("高一",level="年级"))
print(df.xs("A班",level="班级"))
```

```python
# 多层索引的转换
# 堆叠与展开
import pandas as pd
import numpy as np

# 创建宽格式 DataFrame
df = pd.DataFrame(
    np.arange(12).reshape(3, 4),
    index=pd.MultiIndex.from_tuples(
        [("北京", "2024"), ("上海", "2024"), ("广州", "2024")]
    ),
    columns=pd.MultiIndex.from_tuples(
        [("Q1", "营收"), ("Q1", "利润"), ("Q2", "营收"), ("Q2", "利润")]
    )
)
print(df)

# unstack：将内层索引转为列
df_unstacked=df.unstack()
print(df_unstacked)

# stack：将列转为内层索引
df_stacked = df_unstacked.stack()
print(df_stacked)
```
