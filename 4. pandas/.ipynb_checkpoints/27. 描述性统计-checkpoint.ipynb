```python
# 基本统计函数
# 快速获取数据的整体分布情况
import pandas as pd
import numpy as np

# 创建示例数据
df = pd.DataFrame({
    "年龄": [25, 30, 28, 35, 22, 40, 38, 45, 29, 31],
    "薪资": [12000, 15000, 11000, 18000, 9000, 20000, 17000, 22000, 12500, 14000],
    "绩效": [85, 92, 78, 88, 65, 95, 82, 90, 75, 89]
})

print(df.describe())
```

```python
# 计算分位数
import pandas as pd
import numpy as np

s = pd.Series(range(1, 101))  # 1到100

# 最小 0%
print(s.quantile(0))

# 25%
print(s.quantile(0.25))

# 50%
print(s.quantile(0.5))

# 同时计算多个分位数
print(s.quantile([0.1,0.5,0.75,1]))
```

```python
# describe 汇总
# describe() 提供数据集中数值列的综合统计摘要。

import pandas as pd

df = pd.DataFrame({
    "年龄": [25, 30, 28, 35],
    "薪资": [12000, 15000, 11000, 18000],
    "城市": ["北京", "上海", "广州", "深圳"]
})

# 数值列统计
print(df.describe())

# 包含所有列
print(df.describe(include="all"))
```

```python
# 相关系数与协方差
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "年龄": [25, 30, 28, 35, 22],
    "薪资": [12000, 15000, 11000, 18000, 9000],
    "绩效": [85, 92, 78, 88, 65]
})

# 相关系数矩阵
print(df.corr())

# 与特定列的相关
print(df.corr()["薪资"])
```
