```python
# 写入parquet
import pandas as pd
df = pd.DataFrame({
    "id": range(1, 10001),
    "name": ["用户" + str(i) for i in range(1, 10001)],
    "age": [20, 25, 30, 35] * 2500,
    "city": ["北京", "上海", "广州", "深圳"] * 2500,
    "score": [round(i * 0.1, 2) for i in range(1, 10001)]
})

df.to_parquet("data.parquest",index=False)

# 指定压缩方式
df.to_parquet("data.parquest",compression="snappy",index=False)
df.to_parquet("data.parquest",compression="gzip",index=False)
df.to_parquet("data.parquest",compression=None,index=False)
```

```python
# 读取parquet
import pandas as pd

df=pd.read_parquet("data.parquest")
print(df)

# 指定使用特定的引擎
#df=pd.read_parquet("data.parquest",engine="pyarray")
#df=pd.read_parquet("data.parquest",engine="fastparquet")


# 读取指定列
df=pd.read_parquet("data.parquest",columns=["id"])
print(df)
```

```python
# 使用快parquet引擎
import pandas as pd
import time

# 测试不同引擎的写入性能
df = pd.DataFrame({
    "id": range(1, 100001),
    "value": range(1, 100001)
})

df.to_parquet("pyarray.parquet",engine="pyarrow",compression="snappy")
```

```python
# feather
# pip install pyarrow

df = pd.DataFrame({
    "id": range(1, 10001),
    "name": ["用户" + str(i) for i in range(1, 10001)],
    "value": [round(i * 0.1, 2) for i in range(1, 10001)]
})
df.to_feather("data.feather")
# LZ4 压缩 速度快
df.to_feather("data.feather",compression="lz4")
# ZSTD 压缩 压缩率高
df.to_feather("data.feather",compression="zstd")
```

```python
# 读取feather
import pandas as pd

df=pd.read_feather("data.feather")
print(df)

# 读取指定列
df=pd.read_feather("data.feather",columns=["id"])
print(df)
```
