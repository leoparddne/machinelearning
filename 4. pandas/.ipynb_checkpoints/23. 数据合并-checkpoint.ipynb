```python
# merge
import pandas as pd

# 创建两个 DataFrame
df1 = pd.DataFrame({
    "学号": ["S001", "S002", "S003"],
    "姓名": ["张三", "李四", "王五"]
})

df2 = pd.DataFrame({
    "学号": ["S001", "S002", "S003"],
    "数学": [85, 92, 78]
})


print(pd.merge(df1,df2,on="学号"))
```

```python
# 不同列名合并
import pandas as pd

df1 = pd.DataFrame({
    "学号": ["S001", "S002", "S003"],
    "姓名": ["张三", "李四", "王五"]
})

df2 = pd.DataFrame({
    "student_id": ["S001", "S002", "S003"],
    "数学": [85, 92, 78]
})

result=pd.merge(df1,df2,left_on="学号",right_on="student_id")
print(result)

# 删除多余列
result=result.drop("student_id",axis=1)
print(result)
```

```python
# 合并类型
import pandas as pd

df1 = pd.DataFrame({
    "学号": ["S001", "S002", "S003", "S004"],
    "姓名": ["张三", "李四", "王五", "赵六"]
})

df2 = pd.DataFrame({
    "学号": ["S001", "S002", "S003", "S005"],
    "数学": [85, 92, 78, 88]
})

# inner join（默认）
print(pd.merge(df1,df2,on="学号",how="inner"))

print(pd.merge(df1, df2, on="学号", how="left"))
print(pd.merge(df1, df2, on="学号", how="right"))
print(pd.merge(df1, df2, on="学号", how="outer"))
```

```python
# 多键合并
import pandas as pd

df1 = pd.DataFrame({
    "城市": ["北京", "北京", "上海", "上海"],
    "年份": [2023, 2024, 2023, 2024],
    "销售额": [100, 120, 90, 110]
})

df2 = pd.DataFrame({
    "城市": ["北京", "北京", "上海", "上海"],
    "年份": [2023, 2024, 2023, 2024],
    "利润": [20, 25, 18, 22]
})

result=pd.merge(df1,df2,on=["城市","年份"])
print(result)
```

```python
# join 默认左连接，按索引合并
import pandas as pd

# 创建数据，设置索引
df1 = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"],
    "年龄": [25, 30, 28]
}, index=["S001", "S002", "S003"])

df2 = pd.DataFrame({
    "数学": [85, 92, 78],
    "英语": [90, 88, 95]
}, index=["S001", "S002", "S003"])
print(df1.join(df2))
```

```python
# 多表合并
import pandas as pd

df1 = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"]
}, index=["S001", "S002", "S003"])

df2 = pd.DataFrame({
    "数学": [85, 92, 78]
}, index=["S001", "S002", "S004"])

df3 = pd.DataFrame({
    "英语": [90, 88, 95]
}, index=["S001", "S003", "S004"])

result=df1.join(df2,how="left").join(df3,how="left")
print(result)

result2 = df1.join(df2, how="outer").join(df3, how="outer")
print(result2)
```

```python
# concat拼接
import pandas as pd

df1 = pd.DataFrame({
    "姓名": ["张三", "李四"],
    "年龄": [25, 30]
})

df2 = pd.DataFrame({
    "姓名": ["王五", "赵六"],
    "年龄": [28, 35]
})

result=pd.concat([df1,df2],ignore_index=True)
print(result)
```

```python
# 列方向拼接
import pandas as pd

df1 = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"]
})

df2 = pd.DataFrame({
    "年龄": [25, 30, 28],
    "城市": ["北京", "上海", "广州"]
})

result=pd.concat([df1,df2],axis=1)
print(result)
```

```python
# 处理重复索引
import pandas as pd

df1 = pd.DataFrame({
    "姓名": ["张三", "李四"],
    "年龄": [25, 30]
}, index=[0, 1])

df2 = pd.DataFrame({
    "姓名": ["王五", "赵六"],
    "年龄": [28, 35]
}, index=[0, 1])
print(pd.concat([df1, df2]))

# 重新生成索引
print(pd.concat([df1,df2],ignore_index=True))
```

```python
# 验证重复键
import pandas as pd

df1 = pd.DataFrame({
    "A": [1, 2]
})

df2 = pd.DataFrame({
    "A": [3, 4]
})
print(pd.concat([df1, df2], verify_integrity=True,ignore_index=True))
```

```python
# 处理列不匹配
import pandas as pd

df1 = pd.DataFrame({
    "A": [1, 2, 3],
    "B": ["a", "b", "c"]
})

df2 = pd.DataFrame({
    "B": ["x", "y", "z"],
    "C": [10, 20, 30]
})

print(pd.concat([df1,df2],join="outer"))
print(pd.concat([df1,df2],join="inner"))
```

```python
# 只添加新列
import pandas as pd

df1 = pd.DataFrame({
    "姓名": ["张三", "李四"],
    "年龄": [25, 30]
})

df2 = pd.DataFrame({
    "城市": ["北京", "上海"]
})
print(pd.concat([df1,df2],axis=1))
```

```python
# keys 参数创建层次索引
import pandas as pd

df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})
df3 = pd.DataFrame({"A": [9, 10], "B": [11, 12]})

result = pd.concat([df1, df2, df3], keys=["第一年", "第二年", "第三年"])
print(result)


# 从层次索引获取数据
print(result.loc["第二年"])
```
