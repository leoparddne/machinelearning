# DataFrame
DataFrame 是一个表格型的数据结构，含有一组有序的列，每列可以是不同的值类型。  
DataFrame 既有行索引也有列索引，可被看做由 Series 组成的字典（共同用一个索引）。  
DataFrame 是一个二维表格。  

```python
import pandas as pd

data = [['Google', 10], ['Runoob', 12], ['Wiki', 13]]
columns=['Site', 'Age']
df=pd.DataFrame(data,columns=columns)
print(df)

# 单独设置值类型
df["Age"]=df["Age"].astype(float)
print(df)
```

```python
# 字典创建
data={'x':['a','b'],'x1':['c','d']}
df=pd.DataFrame(data)
print(df)

# key为列名
data=[{'a':1,'b':2},{'a':5,'b':10}]
df=pd.DataFrame(data)
print(df)
```

```python
# ndarray创建
import numpy as np
ndarray=np.array([
    ['Google', 10],
    ['Runoob', 12],
    ['Wiki', 13]
])

df=pd.DataFrame(ndarray,columns=['s1','a1'])
print(df)
```

```python
# loc 返回指定行的数据
d=df.loc[0]
print(d)

# 获取多行
d=df.loc[[0,1]]
print(d)
```

```python
# 指定索引名称
df=pd.DataFrame(ndarray,columns=['s1','a1'],index=['a1','a2','a3'])
print(df)
print(df.loc['a2'])
```

```python
# 常用方法
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df=pd.DataFrame(data)
print(df.head(2))
print(df.describe())

df_sorted=df.sort_values(by='Age',ascending=False)
print(df_sorted)

# 选择指定列
print(df[['Name','Age']])

# 按索引选择行
print(df.loc[1:2])

# 分组统计
print(df.groupby('City')['Age'].mean())

# 缺失值填充
df['Age']=df['Age'].fillna(30)
print(df)
```

```python
# 修改数据
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df=pd.DataFrame(data)
df['Name']=['a1','a2','a3','a4']
print(df)

# 增加列
df['NewColumn']=[10,20,20,30]
print(df)

# 删除列
drop=df.drop('City',axis=1)
print(drop)

# 删除行
df_drop=df.drop(0)
print(df_drop)
```

```python
# 聚合函数
df['City'].sum()

print(df)
# 重置索引
df_reset=df.reset_index(drop=True)
print(df_reset)

# 将指定列作为索引
df_set=df.set_index('Name')
print(df_set)

# bool索引
df[df['Age']>25]
```

```python
# 合并、分割
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df1=pd.DataFrame(data)
df2=pd.DataFrame(data)

pd.concat([df1,df2],ignore_index=True)
pd.merge(df1,df2,on='City')
```
