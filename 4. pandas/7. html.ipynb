```python
# 读取多个表格
import pandas as pd

tables=pd.read_html("test.html")
for i,table in enumerate(tables):
    print(f"table {i}")
    print(table.head(3))
```

```python
# 筛选
import pandas as pd

# match需要使用表格内部的文本
tables = pd.read_html("test.html", match="姓名", encoding="utf-8")
print(f"找到 {len(tables)} 个表格")

if tables:
    df = tables[0]
    print(df)
```

```python
# attrs精确定位
import pandas as pd

# match需要使用表格内部的文本
tables = pd.read_html("test.html", attrs={"id":"user"})
print(f"找到 {len(tables)} 个表格")

if tables:
    df = tables[0]
    print(df)
```
