# Series
Series是 Pandas 中的一个核心数据结构，类似于一个一维的数组，具有数据和索引。
Series 可以存储任何数据类型（整数、浮点数、字符串等），并通过标签（索引）来访问元素。  

Series 的大小在创建后是不变的，但可以通过某些操作（如 append 或 delete）来改变。

当对多个 Series 进行运算时，Pandas 会自动根据索引对齐数据。


```python
import pandas as pd

series=pd.Series([1,2,3,4],name="n1")
print(series)
```

```python
# 自定义索引
data=[1,2,3,4,]
custom_index=[1,2,3,4]
series_index=pd.Series(data,index=custom_index,name="n2")
print(series_index)
print(series_index[2])

# 字符串索引
str_index=["a","b","c","d"]
series_str_index=pd.Series(data,index=str_index,name="n3")
print(series_str_index)
print(series_str_index["c"])

# 使用kv创建
obj={1:"a",2:"b",3:"c"}
series_obj=pd.Series(obj)
print(series_obj)
```

```python
# 唯一性说明
# 索引可以是唯一的，但不是必须的
series=pd.Series([1,2,3,4],name="n1")
d=pd.Series(series,)
```

```python
# 读取部分数据
data = [1, 2, 3, 4, 5, 6,7]
index = ['a', 'b', 'c', 'd', 'e', 'f','f']
s = pd.Series(data, index=index)
print(s)
# 一次读到两条相同索引的数据
print(s['f'])
```

```python
# 常用方法
data = [1, 2, 3, 4, 5, 6]
index = ['a', 'b', 'c', 'd', 'e', 'f']
s = pd.Series(data, index=index)

print(s.index)
print(s.values)
print(s.dtype)
print(s.head(2))

# 使用map翻倍元素
double=s.map(lambda f:f*2)
print(double)

# sum
sum=s.sum()
print(sum)

sum2=s.cumsum()
print(sum2)

print(s.isnull())

sortedValue=s.sort_values()
print(sortedValue)
```

```python
data = [1, 2, 3, 4, 5, 6]
index = ['a', 'b', 'c', 'd', 'e', 'f']
s = pd.Series(data, index=index)

# 切片
r=s[3:9]
print(r)

# 添加
s['g']=7
print(s)

# 删除
del s['g']
print(s)

# 删除标签并返回新的series
s_drop=s.drop(['b'])
print(s)
print(s_drop)
```

```python
# 计算
r=s*2
print(r)

r2=s[s>3]
print(r2)
```
