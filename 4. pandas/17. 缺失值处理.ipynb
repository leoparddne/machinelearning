```python
import pandas as pd
import numpy as np

s = pd.Series([1, 2, np.nan, 4, 5])
print(s)

# DataFrame 中的缺失值
df = pd.DataFrame({
    "A": [1, 2, np.nan, 4],
    "B": [np.nan, 2, 3, 4],
    "C": [1, 2, 3, np.nan]
})
print(df)
print(df.isna())
print(df.isna().sum())
```

```python
# dropna 删除缺失值
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, np.nan, 4],
    "B": [1, np.nan, 3, 4],
    "C": [1, 2, 3, np.nan]
})
print(df.dropna())

# how='all'：只有全部为缺失值才删除
print(df.dropna(how="all"))

# thresh：至少有 N 个非缺失值才保留
print(df.dropna(thresh=2))
```

```python
# 删除列
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, np.nan, 4],
    "B": [np.nan, np.nan, np.nan, np.nan],  # 全部缺失
    "C": [1, 2, 3, 4]
})

# 删除包含缺失值的列
print(df.dropna(axis=1))

# how='all'：删除全部为缺失值的列
print(df.dropna(axis=1,how="all"))
```

```python
# fillna 填充缺失值
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, np.nan, 4, 5],
    "B": [np.nan, 2, 3, np.nan, 5],
    "C": [1, 2, 3, 4, np.nan]
})

# 用 0 填充
print(df.fillna(0))

# 用指定值填充不同列
print(df.fillna({"A": 0, "B": 99, "C": -1}))

# 用前一个值填充（前向填充）
print(df.ffill())

# 用后一个值填充（后向填充）
print(df.bfill())
```

```python
# 统计值填充
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, np.nan, 4, 5, 6],
    "B": [10, np.nan, 30, np.nan, 50, 60]
})

# 用均值填充
print(df.fillna(df.mean()))

print("用中位数填充：")
print(df.fillna(df.median()))

# 按列分别填充
print(df.fillna({"A":df["A"].mean(),"B":0}))
```

```python
# interpolate 插值填充
import pandas as pd
import numpy as np

d=[1, 2, np.nan, 4, 5, np.nan, 7]
s = pd.Series(d)

# 线性插值（默认）
print(s.interpolate())

# 指定插值方法
# time 加权插值（时间序列）
s2=pd.Series(d,index=pd.date_range("2024-01-01",periods=len(d),freq="D"))
print(s2.interpolate(method="time"))

# 邻近值填充
print(s.interpolate(method="nearest"))
```

```python
# DataFrame 插值
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, 2, np.nan, 4, 5],
    "B": [10, np.nan, 30, 40, np.nan]
})

# 在原 DataFrame 上插值
df_interpolated =df.interpolate(method="linear")
print(df_interpolated)

# 限制插值范围
# 只填充连续缺失值的首尾（最多1个）
print(df.interpolate(limit=1))
```
