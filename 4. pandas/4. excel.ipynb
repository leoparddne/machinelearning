```python
# 写入excel
df = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie'],
'Age': [25, 30, 35],
'City': ['New York', 'Los Angeles', 'Chicago']
})

df.to_excel("out.xlsx",sheet_name="sheet1",index=False)

df2 = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie'],
'Age': [25, 30, 35],
'City': ['New York', 'Los Angeles', 'Chicago']
})

# 写入多个表单
with pd.ExcelWriter("out2.xlsx") as writer:
    df.to_excel(writer,sheet_name="sheet1",index=False)
    df2.to_excel(writer,sheet_name="sheet2",index=False)
```

```python
# 查看所有表单名称
excelFile=pd.ExcelFile("data.xlsx")
print(excelFile.sheet_names)

# 读取表单
df=excelFile.parse("data")
print(df)

# 关闭文件
excelFile.close()
```

```python
# 设置时间序列化格式
from datetime import date,datetime
df = pd.DataFrame(
    [
        [date(2014, 1, 31), date(1999, 9, 24)],
        [datetime(1998, 5, 26, 23, 33, 4), datetime(2014, 2, 28, 13, 5, 13)],
    ],
    index=["Date", "Datetime"],
    columns=["X", "Y"],
) 

with pd.ExcelWriter("date.xlsx",date_format="YYYY-MM-DD",datetime_format="YYYY-MM-DD HH:MM:SS") as writer:
    df.to_excel(writer)
```
