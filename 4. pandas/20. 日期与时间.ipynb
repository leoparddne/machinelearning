```python
import pandas as pd

# 创建日期范围
dates = pd.date_range("2024-01-01", periods=10, freq="D")

print(dates)

# 按月创建
month=pd.date_range("2024-01-01", periods=12, freq="ME")
print(month)

# 按小时
hour=pd.date_range("2024-01-01", periods=24, freq="h")
print(hour)
```

```python
# DatetimeIndex 与 DatetimeArray
import pandas as pd
import numpy as np

dt = pd.DatetimeIndex([
    pd.Timestamp("2024-01-01"),
    pd.Timestamp("2024-01-02"),
    pd.Timestamp("2024-01-03")
])
print(dt)

# 创建带时间的 Series
s = pd.Series(
    [100, 200, 300],
    index=dt
)
print(s)
```

```python
# 字符串转日期
import pandas as pd

# 各种格式的字符串转日期
dates_str = ["2024-01-01", "2024/01/02", "01-03-2024", "20240104"]

# 自动推断格式
dt2 = pd.to_datetime(dates_str, format="%Y-%m-%d", errors="coerce")
print(dt2)
```

```python
# read_csv 自动解析日期
import pandas as pd
from io import StringIO

# 模拟 CSV 数据
csv_data = """日期,销售额
2024-01-01,100
2024-01-02,200
2024-01-03,150
"""

# 读取后转换
df=pd.read_csv(StringIO(csv_data))
print(df)

df["日期"]=pd.to_datetime(df["日期"])
print(df)
print(df.dtypes)

# 读取时直接解析
df2=pd.read_csv(StringIO(csv_data),parse_dates=["日期"])
print(df2.dtypes)
```

```python
# 日期属性访问
import pandas as pd

s = pd.Series(pd.date_range("2024-01-15", periods=5, freq="D"))
print(s)
print(s.dt.year)
print(s.dt.month)
print(s.dt.day)
print(s.dt.dayofweek)
print(s.dt.day_name())
```

```python
# 日期加减
import pandas as pd

date = pd.Timestamp("2024-01-15")

print(date+pd.Timedelta(days=3))

# 日期差
date1 = pd.Timestamp("2024-01-01")
date2 = pd.Timestamp("2024-01-15")

print((date1-date2))
print((date1-date2).days)

# 日期 Series 运算
dates=pd.date_range("2024-01-01", periods=5, freq="D")
print(dates+pd.Timedelta(days=3))
```

```python
# 日期偏移
import pandas as pd

date = pd.Timestamp("2024-01-15")

# 月初/月末
print(date+pd.offsets.MonthBegin(1))
print(date+pd.offsets.MonthEnd(1))

# 年份偏移
print(date+pd.DateOffset(years=1))
print(date+pd.DateOffset(months=1))

# 工作日
print(date+pd.offsets.Week(weekday=0))
```

```python
# 时区处理
import pandas as pd

# 创建不带时区的时间
dates = pd.date_range("2024-01-01 10:00", periods=3, freq="h")
print(dates)

# 设置时区
dates_utc=dates.tz_localize("UTC")
print(dates_utc)

# 转换时区
date_shanghai=dates_utc.tz_convert("Asia/Shanghai")
print(date_shanghai)
```
