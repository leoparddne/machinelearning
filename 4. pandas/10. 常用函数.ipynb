```python
import pandas as pd

data = [
    {"name": "Google", "likes": 25, "url": "https://www.google.com"},
    {"name": "Runoob", "likes": 30, "url": "https://www.runoob.com"},
    {"name": "Taobao", "likes": 35, "url": "https://www.taobao.com"}
]

df=pd.DataFrame(data)

# 选择指定列
d=df["name"]
print(d)

# 通过标签选择指定的数据
data=df.loc[0,"name"]
print(data)

# 通过指定位置选择数据
data=df.iloc[0,0]
print(data)

# 选择指定的列
data=df.filter(items=["name","likes"])
print(data)

# 正则匹配列名
data=df.filter(regex='like*')
print(data)

# 随机选择n行数据
data=df.sample(n=2)
print(data)
```

```python
# 排序
import pandas as pd

data = [
    {"name": "Runoob", "likes": 25, "url": "https://www.google.com"},
    {"name": "A", "likes": 30, "url": "https://www.runoob.com"},
    {"name": "Taobao", "likes": 35, "url": "https://www.taobao.com"}
]

df=pd.DataFrame(data)

d=df.sort_values("name")
print(d)

# 多个字段排序
d=df.sort_values(["name","likes"])
print(d)

d=df.sort_index()
```

```python
# 分组、聚合
import pandas as pd

data = [
    {"name": "Runoob", "likes": 25, "url": "https://www.google.com"},
    {"name": "A", "likes": 30, "url": "https://www.runoob.com"},
    {"name": "Taobao", "likes": 35, "url": "https://www.taobao.com"}
]

df=pd.DataFrame(data)

d=df.groupby("name")
print(d)

a=d.aggregate("name")
print(a)
```

```python
# 数据合并
import pandas as pd

data = [
    {"name": "Runoob", "likes": 25, "url": "https://www.google.com"},
    {"name": "A", "likes": 30, "url": "https://www.runoob.com"},
    {"name": "Taobao", "likes": 35, "url": "https://www.taobao.com"}
]
df1=pd.DataFrame(data)
df2=pd.DataFrame(data)


d=pd.concat([df1,df2])

d=pd.merge(df1,df2,on='name')
```
