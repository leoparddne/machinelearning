# loc、iloc、at差异
| 特性 | loc | iloc |
|------|-----|------|
| 索引方式 | 标签索引（label-based） | 位置索引（integer-based） |
| 切片 | 包含结束位置 | 不包含结束位置 |
| 单个值 | 返回标量 | 返回标量 |
| 推荐场景 | 有明确索引标签时 | 按位置选择时 |

```python
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六", "钱七"],
    "年龄": [25, 30, 28, 35, 22],
    "城市": ["北京", "上海", "广州", "深圳", "杭州"]
}, index=[1, 3, 5, 7, 9])  # 注意：索引不是连续的

# loc: 使用标签索引（包含结束位置）
print(df.loc[1:5])

# iloc: 使用位置索引（不包含结束位置）
print(df.iloc[0:2])
```

```python
# loc
# 选择行
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六"],
    "年龄": [25, 30, 28, 35],
    "城市": ["北京", "上海", "广州", "深圳"]
}, index=["a", "b", "c", "d"])

# 选择单行（返回 Series）
print(df.loc["a"])

# 选择多行
print(df.loc[["a","c"]])

# 切片选择（包含开始和结束）
print(df.loc["a":"c"])

# 选择列
print(df.loc[:,"姓名"])
print(df.loc[:,["姓名","城市"]])

# 切片选择列
print(df.loc[:,"姓名":"城市"])

# 选择特定行列
print(df.loc["a","姓名"])

# 选择多行多列
print(df.loc[["a","c"],["姓名","城市"]])

# 条件选择
print(df.loc[df["年龄"]>28])
```

```python
# iloc
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六"],
    "年龄": [25, 30, 28, 35],
    "城市": ["北京", "上海", "广州", "深圳"]
})

# 选择行
# 选择第0行
print(df.iloc[0])

# 选择前3行
print(df.iloc[0:3])
print(df.iloc[:3])

# 选择特定行
print(df.iloc[[1,2,3]])

# 负数索引（从末尾开始）
print(df.iloc[-1])

# 选择最后3行
print(df.iloc[-3:])

# 选择列
print(df.iloc[:,0])

print(df.iloc[:,[1,2]])
print(df.iloc[:,0:2])
```

```python
# at 与 iat（获取单个值）
import pandas as pd
import time
import numpy as np

df = pd.DataFrame(np.random.randn(1000, 10), columns=[f"col_{i}" for i in range(10)])
print(df.count())

# at 获取单个值（标签索引）
print(df.at[0,'col_0'])

# iat 获取单个值（位置索引）
print(df.iat[0,0])
```

```python
# 条件选择
import pandas as pd

df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六"],
    "年龄": [25, 30, 28, 35],
    "城市": ["北京", "上海", "广州", "北京"],
    "薪资": [12000, 15000, 11000, 18000]
})

print(df[df["年龄"]>28])

# 多条件
print(df[(df["城市"]=="北京")&(df["薪资"]>12000)])

# isin 筛选
print(df[df["城市"].isin(["北京","上海"])])

# 字符串包含
print(df[df["姓名"].str.contains("李")])
```
