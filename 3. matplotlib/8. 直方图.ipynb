```python
import matplotlib.pyplot as plt
import numpy as np

# 生成一组随机数据
data = np.random.randn(1000)

plt.hist(data,bins=30,color='skyblue',alpha=0.8)

plt.show()
```

```python
# 绘制多个直方图
data1 = np.random.normal(0, 1, 1000)
data2 = np.random.normal(2, 1, 1000)
data3 = np.random.normal(-2, 1, 1000)

plt.hist(data,bins=30,alpha=0.8,label='data')
plt.hist(data2,bins=30,alpha=0.8,label='data2')
plt.hist(data3,bins=30,alpha=0.8,label='data3')

plt.legend()

plt.show()
```

```python
# 结合pandas
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

random_data = np.random.normal(170, 10, 250)

# 转为Pandas DataFrame
dataframe=pd.DataFrame(random_data)

dataframe.hist()

plt.show()
```

```python
data = pd.Series(random_data)

plt.hist(data)
plt.show()
```
