```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y = np.array([1, 4, 9, 16])

plt.title("RUNOOB grid() Test")
plt.xlabel("x - label")
plt.ylabel("y - label")

plt.plot(x, y)

plt.grid()

plt.show()
```

```python
# 只显示x轴网格线

plt.title("RUNOOB grid() Test")
plt.xlabel("x - label")
plt.ylabel("y - label")

plt.plot(x, y)

plt.grid(axis='x')

plt.show()
```

```python
# 设置网格颜色

plt.plot(x, y)

plt.grid(color='r',linestyle='--',linewidth=0.5)

plt.show()
```
