```python
# 点虚线
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([6, 2, 13, 10])
plt.plot(ypoints,linestyle='dotted')
plt.show()
```

```python
# 简写
plt.plot(ypoints,ls='dotted')
plt.show()
```

```python
# 颜色
plt.plot(ypoints,ls='dotted',color='r')
plt.show()
```

```python
# 颜色简写&rgb
plt.plot(ypoints,ls='dotted',c='#8FBC8F')
plt.show()
```

```python
# 线宽
plt.plot(ypoints,ls='dotted',c='#8FBC8F',linewidth='10')
plt.show()
```
