```python
# subplot可以定义整个区域按行列分块
# 子区域按照1-N排序(从左到用、从上到下)
```

```python
# 多图
import matplotlib.pyplot as plt
import numpy as np

#plot 1:
xpoints = np.array([0, 6])
ypoints = np.array([0, 100])

plt.subplot(1, 2, 1)
plt.plot(xpoints,ypoints)
plt.title("plot 1")

#plot 2:
x = np.array([1, 2, 3, 4])
y = np.array([1, 4, 9, 16])

plt.subplot(1, 2, 2)
plt.plot(x,y)
plt.title("plot 2")

plt.suptitle("title")
plt.show()
```

```python
# 定义2x2 四个绘图区域
plt.subplot(2, 2, 1)
plt.plot(xpoints,ypoints)
plt.title("plot 1")

plt.subplot(2, 2, 2)
plt.plot(xpoints,ypoints)
plt.title("plot 2")

plt.subplot(2, 2, 3)
plt.plot(xpoints+5,ypoints)
plt.title("plot 1")

plt.subplot(2, 2, 4)
plt.plot(xpoints,ypoints)
plt.title("plot 1")
plt.show()
```
