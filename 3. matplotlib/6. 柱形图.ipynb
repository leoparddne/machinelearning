```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["Runoob-1", "Runoob-2", "Runoob-3", "C-RUNOOB"])
y = np.array([12, 22, 6, 18])

plt.bar(x,y)
plt.show()
```

```python
#  方向

plt.barh(x,y)
plt.show()
```

```python
# 颜色
plt.barh(x,y,color="#ccc")
plt.show()
```

```python
# 独立设置颜色
color = ["#4CAF50","red","hotpink","#556B2F"]

plt.bar(x,y,color=color)
plt.show()
```

```python
# 宽度
plt.bar(x,y,width=0.1)
plt.show()
```
