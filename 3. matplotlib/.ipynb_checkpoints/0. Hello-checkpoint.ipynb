```python
import numpy as np
from matplotlib import pyplot as plt

x=np.arange(1,11)
y=2*x+5

plt.title("demo ")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.plot(x,y)
plt.show()
```

```python
# 标题与标签定位
import numpy as np
from matplotlib import pyplot as plt

x=np.arange(1,11)
y=2*x+5

plt.title("demo ",loc="left")
plt.xlabel("x axis",loc="left")
plt.ylabel("y axis",loc="top")
plt.plot(x,y)
plt.show()
```

```python
# 中文字体
import numpy as np
from matplotlib import pyplot as plt
import matplotlib

x=np.arange(1,11)
y=2*x+5

zhfont=matplotlib.font_manager.FontProperties(fname="SourceHanSansSC-Bold.otf")

plt.title("中文标题",fontproperties=zhfont)
plt.xlabel("x轴",fontproperties=zhfont)
plt.ylabel("y axis")
plt.plot(x,y)
plt.show()
```

```python
# 获取系统字体
from matplotlib import pyplot as plt
import matplotlib

a=sorted([f.name for f in matplotlib.font_manager.fontManager.ttflist])
print(a)
```

```python
# 用标记替换直线
import numpy as np
from matplotlib import pyplot as plt

x=np.arange(1,11)
y=2*x+5
plt.plot(x,y,"ob")
plt.show()
```

```python
# 正弦波
x = np.arange(0,  3  * np.pi,  0.1) 
y = np.sin(x)
plt.plot(x,y)
plt.show()
```

```python
# subplot 在同一个图绘制多个内容
x = np.arange(0,  3  * np.pi,  0.1) 
y_sin = np.sin(x) 
y_cos = np.cos(x)  

# 建立 subplot 网格，高为 2，宽为 1  
# 激活第一个 subplot
plt.subplot(2,1,1)
plt.plot(x,y_sin)

# 将第二个 subplot 激活，并绘制第二个图像
plt.subplot(2,1,2)
plt.plot(x,y_cos)
plt.show()
```

```python
# 条形图
x =  [5,8,10] 
y =  [12,16,6] 
x2 =  [6,9,11] 
y2 =  [6,15,7] 

plt.bar(x,y,align='center')
plt.bar(x2,y2,color='g',align='center')
plt.show()
```

```python
# histogram 数据频率分布的图形表示

a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
# bin为边界
bin=[0,20,40,60,80,100]
# hist得到每个区间命中的次数,例如0-20区间的数字有3个
hist,bins=np.histogram(a,bins=bin)
print(hist)
print(bins)
```

```python
# 将直方图转为图形
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27]) 
plt.hist(a,bins=bin)
plt.show()
```
