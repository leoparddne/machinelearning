```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])

plt.pie(y)
plt.show()
```

```python
# 设置标签和颜色
labels=['A','B','C','D']
colors=["#d5695d", "#5d8ca8", "#65a479", "#a564c9"]

plt.pie(y,labels=labels,colors=colors)
plt.show()
```

```python
# 突出显示
# 突出显示第二个扇形
explode = (0, 0.1, 0, 0)

plt.pie(y,explode=explode )
plt.show()
```
