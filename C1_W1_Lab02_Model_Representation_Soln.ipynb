```python
w=100
b=100
print(f"w: {w}")
print(f"b: {b}")
```

```python
def compute_model_output(x,w,b):
    m=len(x)
    #按照x的结构初始化一个数组,所有数据设置位0
    f_wb=np.zeros(m)
    print(f_wb)
    
    for i in range(m):
        print(i)
        f_wb[i]=w*x[i]+b
    
    return f_wb

v=compute_model_output([1,3,2],2,2)
print(v)
```

```python
tmp_f_wb=compute_model_output(x_train,w,b)


plt.scatter(x_train,tmp_f_wb,label='out prediction',c='b')

plt.scatter(x_train,y_train,marker='x',c='r',label='actual values')


plt.title("hourse prices")
plt.ylabel("price in 100s of dollars")
plt.xlabel("size (1000 sqft)")
plt.legend()
plt.show
```
