```python
import numpy as np

# 连接
print(np.char.add(['hello'],['world']))
```

```python
# 多重连接

print(np.char.multiply('hhh',3))
```

```python
# 居中
print(np.char.center('Hello',20,fillchar='*'))
```

```python
# 首字母大写
print(np.char.capitalize('hello'))
```

```python
# 每一个单词首字母大写
print(np.char.title('hello world'))
```

```python
print(np.char.lower(['Hello','world']))
print(np.char.upper(['Hello','world']))
```

```python
print(np.char.split('hello,world',sep=','))
print(np.char.splitlines('hello\nworld'))
```

```python
# strip 移除首尾字符
print(np.char.strip('xxhello world  xx','x'))
```

```python
# 指定分隔符来连接数组中的元素或字符串
print(np.char.join(':','xxhello world  xx'))
```

```python
print(np.char.replace('hello world','ll','cc'))
```

```python
# 编解码
a=np.char.encode('hello','cp500')
print(a)

a=np.char.decode(a,'cp500')
print(a)
```
