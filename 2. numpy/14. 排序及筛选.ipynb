```python
import numpy as np

# sort 升序
a = np.array([[3,7],[9,1]])  
print(np.sort(a))
print(np.sort(a,axis=0))
```

```python
# argsort 升序后的索引
print(np.argsort(a))
print(np.argsort(a,axis=0))
```

```python
# lexsort 多个序列排序
nm =  ('raju','anil','ravi','amar') 
dv =  ('f.y.',  's.y.',  's.y.',  'f.y.') 
ind=np.lexsort((dv,nm))
print(ind)

print([nm[i] +","+dv[i] for i in ind])
```

```python
# 复数排序
x=np.array([ 1.+0.j,  2.+0.j,  3.+0.j,  5.+0.j,  6.+0.j])
print(np.sort_complex(x))
```

```python
# partition 分区排序(部分排序)
a = np.array([3, 4, 2, 1])
x=np.partition(a,3)
print(x)


x=np.partition(a,(1,3))
print(x)

arr = np.array([46, 57, 23, 39, 1, 10, 0, 120])
# 找到第三小的值
thiredMin=arr[np.argpartition(arr, 2)[2]]
print(thiredMin)

# 找到第二大的值
thiredMin=arr[np.argpartition(arr, -2)[-2]]
print(thiredMin)

#同时找到第 3 和第 4 小的值
m=arr[np.argpartition(arr,[2,3])]
print(m)
print(m[0])
```

```python
# argmax、argmin
a = np.array([[30,40,70],[80,20,10],[50,90,60]])  
print (np.argmax(a)) 
print (np.argmin(a)) 
print(np.argmax(a,axis=0))
print(np.argmin(a,axis=0))
```

```python
# nonzero 返回输入数组中非零元素的索引
a = np.array([[30,40,0],[0,20,10],[50,0,60]])  
print(np.nonzero(a))
```

```python
# where
x = np.arange(9.).reshape(3,  3)  
# 获取索引
y = np.where(x >  3)  
print(y)

# 获取满足条件的数据
print(x[y])
```

```python
# extract 抽取满足条件的元素
x = np.arange(9.).reshape(3,  3)  
condition=np.mod(x,2)==0
print(x)
print(condition)
print(np.extract(condition,x))
```
