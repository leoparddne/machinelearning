```python
# numpy中的矩阵库：numpy.matlib
# 返回矩阵而不是ndarray
# 矩阵总是二维的
import numpy as np

print(np.matlib.empty((2,2)))
print(np.empty((2,2)))

print(np.matlib.zeros((2,2)))
print(np.matlib.ones((2,2)))

# 对角线为1，其余位置为0
print(np.matlib.eye(n=5,M=5,k=0,dtype=float))

# 单位阵
print(np.matlib.identity(5,dtype=float))

# 随机
print(np.matlib.rand(5,3))
```

```python
# 矩阵与ndarray转换
i=np.matrix('1,2;3,4')
print(i)

j=np.asarray(i)
print(j)

k=np.asmatrix(j)
print(k)
```
