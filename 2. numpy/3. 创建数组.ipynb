```python
# empty
import numpy as np
x=np.empty([3,2,2],dtype=int)
print(x)
```

```python
# zeros
x=np.zeros((3,2))
print(x)

x=np.zeros([3,2])
print(x)

x=np.zeros((3,2),np.dtype('U4'))
print(x)
```

```python
# ones
x=np.ones((2,3),np.int16)
print(x)
```

```python
# zero like
a=np.zeros_like(x)
print(a)
```

```python
# ones like
a=np.ones_like(x)
print(a)
```

```python
# asarray 已经是ndarray时不复制
x=np.asarray(x)
print(x)

t=[1,2,3]
x=np.asarray(t)
print(t)

t=(1,2,3)
x=np.asarray(t)
print(x)


t=[(1,2,3),(1,2,5)]
x=np.asarray(t)
print(x)
```

```python
# frombuffer 把一段“原始字节/内存缓冲区”
# 直接解释成一维 NumPy 数组
# 常用于处理 bytes、bytearray、memoryview、二进制文件、网络数据等
s=b'hello'
x=np.frombuffer(s,dtype='S1')
print(x)
```

```python
# fromiter 从迭代对象中创建
list=range(5)
it=iter(list)

x=np.fromiter(it,dtype=int)
print(x)
```

```python
# arange 生成一个序列数据

x=np.arange(0,10,1)
print(x)

x=np.arange(0,10,0.1,dtype=float)
print(x)
```

```python
# linspace 等差数列
# 开始值、结束值、样本数量
# (根据指定的起点、终点和元素个数，
# 自动计算出均匀的间隔，生成包含 num 个元素的等差数列)
# 从1-10，10个数据
x=np.linspace(1,10,10)
print(x)

# 从1-1，10个数据
x=np.linspace(1,1,10)
print(x)

# 包含结束值
x=np.linspace(1,10,20,endpoint=True)
print(x)

# 不包含结束值
x=np.linspace(1,10,20,endpoint=False)
print(x)
```

```python
# logstop 等比数列
# num 要生成的等步长的样本数量
# 从10^1 - 10^2
x=np.logspace(1,2,num=2)
print(x)

# base为取的底数
#从 2^0 到 2^9，均匀取 10 个数
x=np.logspace(0,9,10,base=2)
print(x)
```
