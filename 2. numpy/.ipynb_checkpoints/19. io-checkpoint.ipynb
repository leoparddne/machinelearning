```python
import numpy as np 

a=np.array([1,2,3,4,5])
np.save("file.npy",a)

# 自动增加npy后缀名
np.save("file2",a)
```

```python
# 读取文件
z=np.load("file.npy")
print(z)
```

```python
# 批量保存
a = np.array([[1,2,3],[4,5,6]])
b = np.arange(0, 1.0, 0.1)
c = np.sin(b)

# 指定保存名称sin_array
np.savez("batch.npz",a,b,sin_array=c)

r=np.load("batch.npz")
print(r.files)
print(r['arr_0'])
print(r['arr_1'])
print(r['sin_array'])
```

```python
# 文本格式保存
a = np.array([1,2,3,4,5]) 
np.savetxt("out.txt",a)

b=np.loadtxt("out.txt")
print(b)

# 保存为整数并设置逗号分隔
np.savetxt("out.txt",a,fmt="%d",delimiter=",")
b=np.loadtxt("out.txt",delimiter=",")
print(b)
```
