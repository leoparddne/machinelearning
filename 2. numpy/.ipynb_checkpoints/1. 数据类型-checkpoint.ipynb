```python
# 指定数据类型
import numpy as np

dt=np.dtype(np.int32)
print(dt)
```

```python
dt=np.dtype(np.int16)
print(dt)
```

```python
#字符串代替数据类型(简写)
#b	布尔型
#i	(有符号) 整型
#u	无符号整型 integer
#f	浮点型
#c	复数浮点型
#m	timedelta（时间间隔）
#M	datetime（日期时间）
#O	(Python) 对象
#S, a	(byte-)字符串
#U	Unicode
#V	原始数据 (void)
dt=np.dtype('i4')
print(dt)
```

```python
dt=np.dtype("i8")
print(dt)
```

```python
# 字节顺序标注
dt=np.dtype('<i4')
print(dt)
```

```python
# 结构化数据类型
dt=np.dtype([('age',np.int8)])
print(dt)
```

```python
# 应用数据类型
dt=np.dtype([('age',np.int8),])
d=np.array([(20,),(30,),(40,)],dtype=dt)
print(d)
```

```python
# 使用字段名获取数据
dt=np.dtype([('age',np.int8),])
d=np.array([(20,),(30,),(40,)],dtype=dt)
print(d["age"])
```

```python
# 定义多个字段
dt=np.dtype([('age',np.int8),('name','U10'),('byteStr','S10')])
print(dt)
d=np.array([(20,"axx","axx"),(30,"b","axx"),(40,"c","axx")],dtype=dt)
print(d)
```
