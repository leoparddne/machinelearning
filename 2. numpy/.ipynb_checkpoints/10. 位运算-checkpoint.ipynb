```python
import numpy as np
arr1=np.array([True,False,True], dtype=bool)
arr2=np.array([False,True,False], dtype=bool)

print(np.bitwise_and(arr1,arr2))
print(np.bitwise_or(arr1,arr2))
print(np.bitwise_xor(arr1,arr2))
print(np.bitwise_not(arr1))

# 取反
arr_invert=np.invert(np.array([1,2],dtype=np.int8))
print(arr_invert)

shift=np.left_shift(5,2)
print(shift)

right_shift=np.right_shift(10,1)
print(right_shift)
```

```python
a,b=13,17

# 转二进制
print(bin(a),bin(b))

# 按位与
# 结果是00001
print(np.bitwise_and(13,17))

print(np.bitwise_or(13,17))
```

```python
# 数组

print(np.bitwise_invert(5))
print(np.invert(np.array([13],dtype=np.int8)))
print(np.binary_repr(13,width=8))
```
