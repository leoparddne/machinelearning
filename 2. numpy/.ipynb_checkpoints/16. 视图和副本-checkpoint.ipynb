```python
# 视图
# numpy 的切片操作返回原数据的视图
# 调用 ndarray 的 view() 函数产生一个视图
# 视图为引用

# 副本
# Python 序列的切片操作，调用deepCopy()函数
# 调用 ndarray 的 copy() 函数产生一个副本
```

```python
import numpy as np
a = np.arange(6)
b = a          # b 引用同一个对象
c = a.copy()   # c 是深拷贝

print(id(a)) 
print(id(b)) 
print(id(c)) 
```

```python
# 视图、浅拷贝
a = np.arange(6).reshape(3,2)  
# a、b自身的id不同
print(id(a)) 
b=a.view()

print(id(b))

print(b)
b=b.reshape((2,3))
print(b)

print(id(b))

# 修改b没有影响a
print(a.shape)

# 修改数据会影响原数据

arr=np.arange(12)

a=arr[3:]
a[1]=999
print(arr)
```

```python
# 副本、深拷贝
a = np.array([[10,10],  [2,3],  [4,5]])  
b=a.copy()
print(b is a)

b[0,0]=9999
print(a)
print(b)
```
