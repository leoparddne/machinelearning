```python
# numpy中的线性代数库：linalg
```

```python
# dot
import numpy as np
import numpy.matlib

a = np.array([[1,2],[3,4]])
b = np.array([[11,12],[13,14]])
print(np.dot(a,b))
```

```python
# vdota 点积
print(np.vdot(a,b))
```

```python
# inner 内积
print(np.inner(a,b))
```

```python
# matmul 矩阵乘积
print(np.matmul(a,b))

a = [[1,0],[0,1]] 
b = [1,2] 
# 二维和一维运算
print(np.matmul(a,b))
```

```python
# 行列式
a = np.array([[1,2], [3,4]]) 
print(np.linalg.det(a))

b = np.array([[6,1,1], [4, -2, 5], [2,8,7]]) 
print(np.linalg.det(b))
```

```python
# solve 求解线性方程组
# x + y + z = 6
# 2y + 5z = -4
# 2x + 5y - z = 27
A=np.matrix("1,1,1;0,2,5;2,5,-1")
B=np.matrix("6;-4;27")
X=np.linalg.inv(A)*B
print(A)
print(X)
```

```python
# inv 计算矩阵的乘法逆矩阵
x = np.array([[1,2],[3,4]]) 
y=np.linalg.inv(x)
print(y)
print(np.dot(x,y))
```
