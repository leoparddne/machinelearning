```python
import numpy as np

a=np.arange(6).reshape(2,3)

for x in np.nditer(a):
    print(x,end=",")
```

```python
for x in np.nditer(a):
    print(x,end=",")

# 使用C语言顺序 行序优先
print("\n")
for x in np.nditer(a.copy(order='C')):
    print(x,end=",")

print("\n")
# Fortran 列序优先
for x in np.nditer(a.copy(order='F')):
    print(x,end=",")
```

```python
# 直接设置nditer排序
for x in np.nditer(a,order='C'):
    print(x,end=',')
```

```python
# 修改元素
# 默认为只读 (readwrite 或者 writeonly)
for x in np.nditer(a,op_flags=['readwrite']):
    x[...]=2*x

print(a)
```

```python
# 外部循环 - 尽可能将内存中连续的块一次性读出
a = np.arange(0,60,5) 
a = a.reshape(3,4)  

print(a)
print("\n")

# 默认是按行存储,此时按行读就能一次性全部读出
print(a.flags.c_contiguous)  # True，说明是 C 风格（行优先）存储
print(a.flags.f_contiguous)  # False，不是 Fortran 风格
for x in np.nditer(a,flags=['external_loop'], order =  'F'):
    print(x,end=',')

print("\n")
# 每列在内存中不连续,就会分成四块
for x in np.nditer(a,flags=['external_loop'], order =  'C'):
    print(x,end=',')

print("\n")

print(a.flags.c_contiguous)  # True，说明是 C 风格（行优先）存储
print(a.flags.f_contiguous)  # False，不是 Fortran 风格

#转成按列存储
a=np.asfortranarray(a)

# 已经转成按列存储,此时按列读就能一次性全部读出
for x in np.nditer(a,flags=['external_loop'], order =  'F'):
    print(x,end=',')

print("\n")
for x in np.nditer(a,flags=['external_loop'], order =  'C'):
    print(x,end=',')
```

```python
# 广播迭代
a=np.arange(0,60,5)
a=a.reshape(3,4)

b=np.array([1,2,3,4],dtype=int)

for x,y in np.nditer([a,b]):
    print("%d:%d"%(x,y),end=',')
```
