```python
import numpy as np
```

```python
# axis 压缩数据 沿指定方向对数据求和后降低一个维度

a=np.array([[1,2,3],[4,5,6]])
print(a.sum(axis=0))

print(a.sum(axis=1))
```

```python
# 获取维度的数量
print(a.ndim)
```

```python
# 获取数组的维度(得到的元组的长度是维度的数目，得到的数据为具体的维度的数据个数)
a=np.array([[1,2],[3,4],[5,6]])
print(a.shape)
```

```python
# 调整数组大小 - sharp(已弃用,赋值会原地修改数组 a 本身)
a=np.array([[1,2],[3,4],[5,6]])
print(a)

a.shape=(2,3)
print(a)
```

```python
# reshape
a=np.array([[1,2],[3,4],[5,6]])
print(a)
b=a.reshape(2,3)
print(b)
```

```python
# 获取总元素个数
a=np.array([[1,2],[3,4],[5,6]])
print(a.size)
```

```python
# 获取数据类型
a=np.array([[1,2],[3,4],[5,6]])
print(a.dtype)
```

```python
# 指定数据创建时的 数据类型
a=np.array([[1,2],[3,4],[5,6]],dtype=np.int8)
print(a.dtype)
```

```python
# 虚数
a=np.array([[1+4j,2],[3,4],[5,6]])
print(a)
print(a.real)
print(a.imag)
```

```python
# 元素的大小
a=np.array([[1,2],[3,4],[5,6]])
print(a.itemsize)
```

```python
# 对象内存信息(数据排列方式等)
a=np.array([[1,2],[3,4],[5,6]])
print(a.flags)
```
