```python
import numpy as np 
 
a = np.array([1,2,3,4]) 
b = np.array([10,20,30,40]) 

print(a.shape)
print(b.shape)

print(a*b)
print(a+b)
```

```python
# 形状不同时触发广播
a = np.array([[ 0, 0, 0],
           [10,10,10],
           [20,20,20],
           [30,30,30]])
b = np.array([0,1,2])
print(a + b)
```
